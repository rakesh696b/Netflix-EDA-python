```python
import pandas as pd

url = "https://raw.githubusercontent.com/rfordatascience/tidytuesday/master/data/2021/2021-04-20/netflix_titles.csv"
df = pd.read_csv(url)

df.head()
```

```python
# Convert date_added to datetime
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')

# Extract year from date_added
df['year_added'] = df['date_added'].dt.year

# Extract month from date_added
df['month_added'] = df['date_added'].dt.month

df[['date_added', 'year_added', 'month_added']].head()
```

```python
import matplotlib.pyplot as plt

df['type'].value_counts().plot(kind='bar')
plt.title('Count of Movies and TV shows on Netflix')
plt.xlabel('Type')
plt.ylabel('Count')
plt.show()
```

```python
df['country'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 countries with Most content on Netflix")
plt.xlabel("Country")
plt.ylabel("Count")
plt.show()
```

```python
df['year_added'].value_counts().sort_index().plot(kind = 'line')
plt.title("Content Added on Netflix over 10 years")
plt.xlabel("Year")
plt.ylabel("number of Titles added")
plt.show()
```

```python
df['year_added'].value_counts().idxmax()
```

```python
content_trend = df.groupby(['year_added', 'type']).size().unstack(fill_value = 0)
content_trend.plot(kind = 'line')
plt.title("Movies vs TV Shows Added over the Years")
plt.xlabel("Year")
plt.ylabel("Count")
plt.show()
```

```python
genre_counts = df['listed_in'].str.split(', ', expand = True).stack().value_counts().head(10)
genre_counts
```

```python
actor_counts = df['cast'].dropna().str.split(',', expand = True).stack().value_counts().head(10)
actor_counts
```

```python
country_counts = df['country'].dropna().str.split(',', expand = True).stack().value_counts().head(10)
country_counts
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

all_genres = ' '.join(df['listed_in'].dropna())

wordcloud = WordCloud(width = 800, height = 400).generate(all_genres)

plt.figure(figsize = (10,5))
plt.imshow(wordcloud, interpolation = 'bilinear')
plt.axis('off')
plt.show()
```

```python
director_counts = df['director'].dropna().str.split(',', expand = True).stack().value_counts().head(10)
director_counts
```

```python
movies  = df[df['type'] == 'Movie']
tv_shows = df[df['type'] == 'TV Show']

movies['duration_int'] =  movies['duration'].str.replace('min', '').astype(int)

tv_shows['duration_int'] = tv_shows['duration'].str.replace('Seasons', '').str.replace('Season', '').astype(int)

avg_movie_duration = movies['duration_int'].mean()
avg_tv_seasons = tv_shows['duration_int'].mean()

avg_movie_duration, avg_tv_seasons
```
